```python

import pandas as pd
import numpy as np

# Create student dataset
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Student_Name": [
        "Arun", "Priya", "Kavin", "Divya", "Rahul",
        "Meena", "Vijay", "Anu", "Sanjay", "Nila"
    ],
    "Python": [85, 90, 78, 92, 65, 88, 72, 95, 80, 89],
    "Mathematics": [80, 95, 75, 89, 70, 85, 68, 98, 82, 91],
    "English": [78, 88, 82, 90, 60, 92, 75, 94, 85, 87],
    "Attendance": [90, 95, 80, 98, 70, 92, 75, 99, 85, 93]
}

df = pd.DataFrame(data)

# Display dataset
print(df)
```

```python

df.to_csv("student_performance.csv", index=False)

print("Dataset saved successfully!")
```

```python

# Calculate total marks
df["Total"] = df["Python"] + df["Mathematics"] + df["English"]

print(df[["Student_Name", "Total"]])
```

```python

# Calculate average marks
df["Average"] = df[["Python", "Mathematics", "English"]].mean(axis=1)

print(df[["Student_Name", "Average"]])
```

```python

# Assign grades
def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Average"].apply(calculate_grade)

print(df[["Student_Name", "Average", "Grade"]])
```

```python

# Highest and lowest performers
highest = df.loc[df["Average"].idxmax()]
lowest = df.loc[df["Average"].idxmin()]

print("Highest Performer:", highest["Student_Name"])
print("Highest Average:", highest["Average"])

print("Lowest Performer:", lowest["Student_Name"])
print("Lowest Average:", lowest["Average"])
```

```python

# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python

# Check duplicate records
print("Number of duplicate records:", df.duplicated().sum())
```

```python

# Fill missing numerical values
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)

print("Missing values handled successfully!")
```

```python

# Remove duplicate records
df = df.drop_duplicates()

print("Duplicate records removed!")
```

```python

# Final dataset verification
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nCleaned Student Dataset:")
print(df)
```

```python

import matplotlib.pyplot as plt

print("Matplotlib imported successfully!")
```

```python

plt.figure(figsize=(10, 5))

plt.bar(df["Student_Name"], df["Python"], label="Python")
plt.bar(df["Student_Name"], df["Mathematics"], label="Mathematics", alpha=0.7)
plt.bar(df["Student_Name"], df["English"], label="English", alpha=0.5)

plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Student Subject Performance")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

grade_counts = df["Grade"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    grade_counts,
    labels=grade_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Grade Distribution")
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.plot(
    df["Student_Name"],
    df["Average"],
    marker="o",
    linestyle="-"
)

plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.title("Student Average Performance")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.bar(df["Student_Name"], df["Attendance"])

plt.xlabel("Student Name")
plt.ylabel("Attendance (%)")
plt.title("Student Attendance Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Calculate overall class average

class_average = df["Average"].mean()

print("Overall Class Average:", round(class_average, 2))
```

```python

# Find top 3 students

top_students = df.nlargest(3, "Average")

print("Top 3 Performing Students:")
print(top_students[["Student_Name", "Average", "Grade"]])
```

```python

# Calculate subject-wise average

subject_average = df[["Python", "Mathematics", "English"]].mean()

print("Subject-wise Average Marks:")
print(subject_average.round(2))
```

```python

# Count students in each grade

grade_distribution = df["Grade"].value_counts()

print("Grade Distribution:")
print(grade_distribution)
```

```python

# Attendance analysis

average_attendance = df["Attendance"].mean()

print("Average Attendance:", round(average_attendance, 2), "%")

low_attendance = df[df["Attendance"] < 75]

print("\nStudents with attendance below 75%:")
print(low_attendance[["Student_Name", "Attendance"]])
```

```python

# Final project summary

print("===== STUDENT PERFORMANCE REPORT =====")

print("Total Students:", len(df))

print("Class Average:", round(df["Average"].mean(), 2))

print("Highest Average:",
      df["Average"].max())

print("Lowest Average:",
      df["Average"].min())

print("Average Attendance:",
      round(df["Attendance"].mean(), 2), "%")

print("\nGrade Distribution:")
print(df["Grade"].value_counts())

print("\n===== END OF REPORT =====")
```
